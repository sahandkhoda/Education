This little example would download a small fraction of the physiological signals from one patient in MIMIC-IV waveform database.

* https://physionet.org/content/mimic4wdb/0.1.0/
* https://physionet.org/files/mimic4wdb/0.1.0/waves/p124/p12408438/88826653/

To use it you need to first install Visual Code, Python with matplotlib. And then wfdb by running this command in the command prompt:

`pip install wfdb`

```python
import matplotlib.pyplot as plt
import wfdb
```

You can change patient and record and which sample to use to find other data:

```python
PATIENT = 'p12408438'
RECORD = '88826653'
SAMPFROM = 9000
```

```python
rec = wfdb.rdrecord(RECORD, pn_dir='mimic4wdb/0.1.0/waves/' + PATIENT[:4] + '/' + PATIENT + '/' + RECORD,
                    sampfrom=SAMPFROM, sampto=SAMPFROM+70,
                    smooth_frames=False)

if not 'Pleth' in rec.sig_name:
    raise ValueError('pleth not found in record')
```

```python
i = rec.sig_name.index('Pleth')

print("N", len(rec.e_p_signal[i]))
print("data", repr(rec.e_p_signal[i]))

plt.plot(rec.e_p_signal[i])
plt.show()
```
